## Task 0: Prepare reusable columns and helper functions

```python
import re
import warnings
from collections import Counter

from sklearn.model_selection import TimeSeriesSplit, cross_val_score
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, accuracy_score, classification_report
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Re-read the source to ensure the advanced analyses start from the original cleaned dataset.
df = pd.read_csv("cleaned_upwork_jobs.csv")
df.columns = [str(c).strip() for c in df.columns]

def first_existing(candidates, columns=None):
    columns = df.columns if columns is None else columns
    lookup = {str(c).lower().strip(): c for c in columns}
    for candidate in candidates:
        if candidate.lower() in lookup:
            return lookup[candidate.lower()]
    return None

TITLE_COL = first_existing(["title", "job_title", "job title", "name"])
DATE_COL = first_existing(["published_date", "date_posted", "posted_date", "created_at", "publication_date", "date"])
COUNTRY_COL = first_existing(["country", "client_country", "location_country", "client location", "location"])
LOW_COL = first_existing(["hourly_low", "hourly_rate_low", "min_hourly_rate", "hourly_min"])
HIGH_COL = first_existing(["hourly_high", "hourly_rate_high", "max_hourly_rate", "hourly_max"])
RATE_COL = first_existing(["average_hourly_rate", "hourly_rate", "rate"])
BUDGET_COL = first_existing(["budget", "fixed_budget", "project_budget"])
DESC_COL = first_existing(["description", "job_description", "details"])

if DATE_COL:
    df[DATE_COL] = pd.to_datetime(df[DATE_COL], errors="coerce", utc=True)

# Construct a numeric hourly rate from the available fields where possible.
if LOW_COL:
    df[LOW_COL] = pd.to_numeric(df[LOW_COL], errors="coerce")
if HIGH_COL:
    df[HIGH_COL] = pd.to_numeric(df[HIGH_COL], errors="coerce")
if RATE_COL:
    df[RATE_COL] = pd.to_numeric(df[RATE_COL], errors="coerce")
if LOW_COL and HIGH_COL:
    df["analysis_hourly_rate"] = df[[LOW_COL, HIGH_COL]].mean(axis=1)
elif RATE_COL:
    df["analysis_hourly_rate"] = df[RATE_COL]
else:
    df["analysis_hourly_rate"] = np.nan

if BUDGET_COL:
    df[BUDGET_COL] = pd.to_numeric(df[BUDGET_COL], errors="coerce")

print("Detected columns")
print({
    "title": TITLE_COL, "date": DATE_COL, "country": COUNTRY_COL,
    "hourly_low": LOW_COL, "hourly_high": HIGH_COL,
    "hourly_rate": RATE_COL, "budget": BUDGET_COL, "description": DESC_COL
})
print("Dataset shape:", df.shape)
```

## Task 1: Analyze job-title keywords and offered hourly rates

The notebook extracts common keywords from job titles and compares their hourly rates. It reports sample size, mean, median, and a Spearman rank correlation between keyword presence (0/1) and hourly rate. Correlation is an association, not proof that a keyword causes higher pay.

```python
if TITLE_COL and df["analysis_hourly_rate"].notna().any():
    task1 = df[[TITLE_COL, "analysis_hourly_rate"]].copy()
    task1[TITLE_COL] = task1[TITLE_COL].fillna("").astype(str).str.lower()
    task1 = task1[task1["analysis_hourly_rate"] > 0].copy()

    stop_words = {"the", "and", "for", "with", "from", "you", "your", "are", "our", "job", "looking", "need", "expert", "work"}
    token_lists = task1[TITLE_COL].apply(lambda x: re.findall(r"[a-z][a-z0-9+#.-]{1,}", x))
    keyword_counts = Counter(word for tokens in token_lists for word in set(tokens) if word not in stop_words)

    rows = []
    for word, count in keyword_counts.items():
        if count >= 5:
            present = token_lists.apply(lambda tokens: word in tokens)
            values = task1.loc[present, "analysis_hourly_rate"].dropna()
            absent_values = task1.loc[~present, "analysis_hourly_rate"].dropna()
            if len(values) >= 5 and len(absent_values) >= 5:
                binary = present.loc[task1["analysis_hourly_rate"].notna()].astype(int)
                rates = task1.loc[task1["analysis_hourly_rate"].notna(), "analysis_hourly_rate"]
                rho = binary.corr(rates, method="spearman")
                rows.append({
                    "keyword": word, "jobs_with_keyword": len(values),
                    "mean_hourly_rate": values.mean(), "median_hourly_rate": values.median(),
                    "mean_rate_without_keyword": absent_values.mean(),
                    "mean_rate_difference": values.mean() - absent_values.mean(),
                    "spearman_correlation": rho
                })
    keyword_salary = pd.DataFrame(rows)
    if not keyword_salary.empty:
        keyword_salary = keyword_salary.sort_values("jobs_with_keyword", ascending=False)
        display(keyword_salary.head(30).round(3))
        plot_df = keyword_salary.sort_values("mean_hourly_rate", ascending=False).head(15)
        plt.figure(figsize=(12, 6))
        sns.barplot(data=plot_df, x="mean_hourly_rate", y="keyword")
        plt.title("Average Hourly Rate by Job-Title Keyword (minimum 5 jobs)")
        plt.xlabel("Average hourly rate (dataset currency)")
        plt.ylabel("Keyword")
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(12, 5))
        sns.barplot(data=keyword_salary.assign(abs_corr=keyword_salary["spearman_correlation"].abs())
                    .sort_values("abs_corr", ascending=False).head(15),
                    x="spearman_correlation", y="keyword")
        plt.axvline(0, color="black", linewidth=1)
        plt.title("Spearman Association: Keyword Presence vs Hourly Rate")
        plt.xlabel("Spearman correlation")
        plt.ylabel("Keyword")
        plt.tight_layout()
        plt.show()
        keyword_salary.to_csv("task1_keyword_salary_analysis.csv", index=False)
    else:
        print("Not enough title keywords have at least five observations with valid hourly rates.")
else:
    print("Task 1 skipped: a job title column and numeric hourly rates are required.")
```

## Task 2: Identify emerging job categories from posting frequency

Job titles are grouped into broad, editable keyword-based categories. Monthly posting counts and the percentage change from the first to last observed month are reported. A category is only labelled “emerging” when the dataset contains at least two months; this is a descriptive signal, not proof of a long-term industry trend.

```python
def classify_title(title):
    t = str(title).lower()
    rules = {
        "AI / Machine Learning": ["machine learning", "artificial intelligence", " ai ", "llm", "generative ai", "prompt engineer"],
        "Data / Analytics": ["data analyst", "data scientist", "data engineer", "analytics", "power bi", "tableau", "sql"],
        "Software / Web Development": ["developer", "software", "programmer", "web development", "frontend", "front-end", "backend", "back-end", "react", "python"],
        "Design / Creative": ["designer", "graphic design", "ui/ux", "ux ", "video editor", "illustrator", "animation"],
        "Marketing / SEO": ["marketing", "seo", "social media", "advertising", "content writer", "copywriter"],
        "Admin / Virtual Assistance": ["virtual assistant", "administrative", "data entry", "customer support", "customer service"],
        "Writing / Translation": ["writer", "writing", "translation", "translator", "proofread", "editor"],
        "Finance / Accounting": ["accountant", "accounting", "bookkeeping", "financial analyst", "finance"],
        "Sales / Business": ["sales", "business development", "lead generation", "recruiter", "project manager"]
    }
    padded = " " + t + " "
    for category, keywords in rules.items():
        if any(k in t for k in keywords):
            return category
    return "Other / Unclassified"

if TITLE_COL and DATE_COL and df[DATE_COL].notna().any():
    trend_df = df[[TITLE_COL, DATE_COL]].dropna().copy()
    trend_df["category"] = trend_df[TITLE_COL].apply(classify_title)
    trend_df["month"] = trend_df[DATE_COL].dt.to_period("M").astype(str)
    monthly_categories = trend_df.groupby(["month", "category"]).size().rename("job_count").reset_index()
    category_wide = monthly_categories.pivot(index="month", columns="category", values="job_count").fillna(0).sort_index()
    display(category_wide.tail(18))

    summary = []
    for category in category_wide.columns:
        series = category_wide[category]
        first, last = series.iloc[0], series.iloc[-1]
        growth = ((last - first) / first * 100) if first > 0 else np.nan
        # Compare first and second halves to reduce sensitivity to a single month.
        midpoint = max(1, len(series) // 2)
        first_half_mean = series.iloc[:midpoint].mean()
        second_half_mean = series.iloc[midpoint:].mean() if midpoint < len(series) else np.nan
        half_growth = ((second_half_mean - first_half_mean) / first_half_mean * 100
                       if pd.notna(second_half_mean) and first_half_mean > 0 else np.nan)
        summary.append({"category": category, "total_postings": int(series.sum()),
                        "first_month_count": int(first), "last_month_count": int(last),
                        "first_to_last_growth_pct": growth,
                        "second_half_vs_first_half_pct": half_growth})
    emerging_categories = pd.DataFrame(summary).sort_values(
        "second_half_vs_first_half_pct", ascending=False, na_position="last")
    display(emerging_categories.round(2))
    category_wide.plot(figsize=(14, 7), marker="o")
    plt.title("Monthly Upwork Job Postings by Broad Category")
    plt.xlabel("Month")
    plt.ylabel("Number of postings")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
    emerging_categories.to_csv("task2_emerging_categories.csv", index=False)
    category_wide.to_csv("task2_monthly_category_counts.csv")
else:
    print("Task 2 skipped: job title and valid publication date columns are required.")
```

## Task 3: Forecast high-demand job roles

This section forecasts total monthly postings by broad category using lagged monthly counts and a Random Forest regressor. It uses a chronological holdout where possible and reports MAE, RMSE, and R². At least 8 monthly observations are recommended; with fewer months, results are not statistically dependable. This is a baseline forecasting exercise, not a guarantee of future demand.

```python
if TITLE_COL and DATE_COL and df[DATE_COL].notna().any():
    forecast_base = df[[TITLE_COL, DATE_COL]].dropna().copy()
    forecast_base["category"] = forecast_base[TITLE_COL].apply(classify_title)
    forecast_base["month"] = forecast_base[DATE_COL].dt.to_period("M").dt.to_timestamp()
    monthly = forecast_base.groupby(["month", "category"]).size().rename("count").reset_index()

    forecast_results = []
    forecast_lines = []
    for category, group in monthly.groupby("category"):
        s = group.set_index("month")["count"].sort_index().asfreq("MS", fill_value=0)
        if len(s) < 8 or s.sum() < 8:
            continue
        supervised = pd.DataFrame({"y": s})
        for lag in [1, 2, 3]:
            supervised[f"lag_{lag}"] = supervised["y"].shift(lag)
        supervised["month_num"] = supervised.index.month
        supervised["time_index"] = np.arange(len(supervised))
        supervised = supervised.dropna()
        if len(supervised) < 6:
            continue
        features = ["lag_1", "lag_2", "lag_3", "month_num", "time_index"]
        split = max(1, int(len(supervised) * 0.8))
        train, test = supervised.iloc[:split], supervised.iloc[split:]
        if test.empty or len(train) < 4:
            continue
        model = RandomForestRegressor(n_estimators=250, random_state=42, min_samples_leaf=1)
        model.fit(train[features], train["y"])
        pred = np.maximum(0, model.predict(test[features]))
        mae = mean_absolute_error(test["y"], pred)
        rmse = np.sqrt(mean_squared_error(test["y"], pred))
        r2 = r2_score(test["y"], pred) if len(test) > 1 else np.nan
        forecast_results.append({"category": category, "test_months": len(test),
                                 "MAE": mae, "RMSE": rmse, "R2": r2,
                                 "observed_total": test["y"].sum(),
                                 "predicted_total": pred.sum()})

        # Recursive one-month-ahead forecast for the next three months.
        history = list(s.values.astype(float))
        future_dates = pd.date_range(s.index.max() + pd.offsets.MonthBegin(1), periods=3, freq="MS")
        for future_date in future_dates:
            row = pd.DataFrame([{
                "lag_1": history[-1], "lag_2": history[-2], "lag_3": history[-3],
                "month_num": future_date.month, "time_index": len(history)
            }])
            yhat = max(0, float(model.predict(row[features])[0]))
            forecast_lines.append({"month": future_date, "category": category,
                                   "forecast_postings": yhat})
            history.append(yhat)

    if forecast_results:
        forecast_metrics = pd.DataFrame(forecast_results).sort_values("predicted_total", ascending=False)
        display(forecast_metrics.round(2))
        forecast_df = pd.DataFrame(forecast_lines)
        if not forecast_df.empty:
            display(forecast_df.round(1))
            plt.figure(figsize=(14, 7))
            for category in forecast_df["category"].unique():
                actual = monthly[monthly["category"] == category].sort_values("month")
                future = forecast_df[forecast_df["category"] == category].sort_values("month")
                plt.plot(actual["month"], actual["count"], marker="o", label=f"{category} actual")
                plt.plot(future["month"], future["forecast_postings"], marker="x", linestyle="--",
                         label=f"{category} forecast")
            plt.title("Monthly Job Posting Counts and 3-Month Baseline Forecast")
            plt.xlabel("Month")
            plt.ylabel("Posting count")
            plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
            plt.tight_layout()
            plt.show()
        forecast_metrics.to_csv("task3_forecast_metrics.csv", index=False)
        forecast_df.to_csv("task3_three_month_forecast.csv", index=False)
    else:
        print("Not enough monthly history for the baseline forecasting model. Aim for 8+ months per category.")
else:
    print("Task 3 skipped: job title and valid publication date columns are required.")
```

## Task 4: Compare average hourly rates by country

Rates are calculated from the available hourly low/high columns or a numeric hourly-rate column. The comparison includes sample size and median to reduce the risk of over-interpreting countries with only a few observations. Currency and rate units are assumed to be consistent in the source data; do not compare currencies unless they have been converted to a common currency.

```python
if COUNTRY_COL and df["analysis_hourly_rate"].notna().any():
    country_rates = df[[COUNTRY_COL, "analysis_hourly_rate"]].copy()
    country_rates[COUNTRY_COL] = country_rates[COUNTRY_COL].fillna("Unknown").astype(str).str.strip()
    country_rates = country_rates[(country_rates["analysis_hourly_rate"] > 0) &
                                  (country_rates[COUNTRY_COL] != "")]
    country_summary = country_rates.groupby(COUNTRY_COL)["analysis_hourly_rate"].agg(
        job_count="count", average_hourly_rate="mean", median_hourly_rate="median"
    ).reset_index()
    # Filter tiny samples for the chart; retain the full table in the saved file.
    country_summary = country_summary.sort_values("average_hourly_rate", ascending=False)
    display(country_summary.head(30).round(2))
    chart_data = country_summary[country_summary["job_count"] >= 3].head(20)
    if not chart_data.empty:
        plt.figure(figsize=(12, 8))
        sns.barplot(data=chart_data, x="average_hourly_rate", y=COUNTRY_COL)
        plt.title("Average Hourly Rate by Country (at least 3 records)")
        plt.xlabel("Average hourly rate (source currency)")
        plt.ylabel("Country")
        plt.tight_layout()
        plt.show()
    else:
        print("No country has at least three valid hourly-rate observations.")
    country_summary.to_csv("task4_hourly_rates_by_country.csv", index=False)
else:
    print("Task 4 skipped: country and numeric hourly-rate columns are required.")
```

## Task 5: Job recommendation engine prototype

The prototype uses TF-IDF text similarity on job titles and, when available, descriptions. Enter skills or a short job preference below; it returns the most similar postings. This content-based recommender does not learn from clicks or past user behaviour. The optional API example below can be run separately after saving the data and installing FastAPI.

```python
if TITLE_COL:
    rec_df = df.copy()
    rec_df[TITLE_COL] = rec_df[TITLE_COL].fillna("").astype(str)
    text_parts = rec_df[TITLE_COL].str.strip()
    if DESC_COL:
        text_parts = text_parts + " " + rec_df[DESC_COL].fillna("").astype(str)
    rec_df["recommendation_text"] = text_parts.str.strip()
    rec_df = rec_df[rec_df["recommendation_text"].str.len() > 0].reset_index(drop=True)
    if len(rec_df) >= 2:
        vectorizer = TfidfVectorizer(stop_words="english", max_features=20000, ngram_range=(1, 2))
        job_matrix = vectorizer.fit_transform(rec_df["recommendation_text"])
        def recommend_jobs(preferences, top_n=10):
            """Return jobs ranked by text similarity to the user's skills/preferences."""
            if not str(preferences).strip():
                raise ValueError("Please enter skills or job preferences.")
            user_vector = vectorizer.transform([str(preferences)])
            scores = cosine_similarity(user_vector, job_matrix).ravel()
            result = rec_df.copy()
            result["similarity_score"] = scores
            show_cols = [TITLE_COL]
            for col in [COUNTRY_COL, DATE_COL, "analysis_hourly_rate", BUDGET_COL]:
                if col and col in result.columns and col not in show_cols:
                    show_cols.append(col)
            show_cols.append("similarity_score")
            return result.sort_values("similarity_score", ascending=False)[show_cols].head(top_n)
        print("Recommendation engine ready. Example:")
        display(recommend_jobs("Python data analysis machine learning SQL", top_n=10))
    else:
        print("Not enough non-empty job titles to build recommendations.")
else:
    print("Recommendation engine skipped: job title column not found.")
```

## Task 6: Track job-market dynamics by month

This section creates a monthly KPI table and charts for posting volume, hourly rate, and fixed-budget amounts where the relevant columns exist. The generated CSV can feed a dashboard. To update it monthly, add newly collected records to the source CSV and rerun the notebook; the notebook itself does not automatically fetch fresh Upwork data.

```python
if DATE_COL and df[DATE_COL].notna().any():
    dated = df[df[DATE_COL].notna()].copy()
    dated["month"] = dated[DATE_COL].dt.to_period("M").astype(str)
    monthly_kpis = dated.groupby("month").size().rename("job_postings").to_frame()
    if "analysis_hourly_rate" in dated:
        rate_kpi = dated.groupby("month")["analysis_hourly_rate"].agg(
            average_hourly_rate="mean", median_hourly_rate="median", hourly_rate_records="count")
        monthly_kpis = monthly_kpis.join(rate_kpi)
    if BUDGET_COL:
        monthly_kpis = monthly_kpis.join(dated.groupby("month")[BUDGET_COL].agg(
            average_budget="mean", median_budget="median", budget_records="count"))
    display(monthly_kpis.round(2))
    monthly_kpis.to_csv("task6_monthly_market_dashboard_data.csv")
    monthly_kpis["job_postings"].plot(figsize=(13, 5), marker="o")
    plt.title("Monthly Upwork Job Posting Volume")
    plt.xlabel("Month")
    plt.ylabel("Number of postings")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
    if "average_hourly_rate" in monthly_kpis.columns:
        monthly_kpis["average_hourly_rate"].plot(figsize=(13, 5), marker="o")
        plt.title("Average Hourly Rate by Month")
        plt.xlabel("Month")
        plt.ylabel("Average hourly rate (source currency)")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
else:
    print("Task 6 skipped: valid publication dates are required.")
```

## Task 7: Investigate remote-work trends

Remote status is inferred from title and description keywords, so this is a proxy rather than a verified contract attribute. Compare remote-keyword and non-remote-keyword posting counts by month. Search terms can be edited to match the dataset's language.

```python
if DATE_COL and df[DATE_COL].notna().any():
    text_cols = [c for c in [TITLE_COL, DESC_COL] if c]
    if text_cols:
        searchable = df[text_cols].fillna("").astype(str).agg(" ".join, axis=1).str.lower()
        remote_pattern = r"\b(remote|work from home|wfh|anywhere|distributed team|fully remote)\b"
        remote_df = df.loc[df[DATE_COL].notna(), [DATE_COL]].copy()
        remote_df["remote_mentioned"] = searchable.loc[remote_df.index].str.contains(remote_pattern, regex=True)
        remote_df["month"] = remote_df[DATE_COL].dt.to_period("M").astype(str)
        remote_monthly = remote_df.groupby(["month", "remote_mentioned"]).size().unstack(fill_value=0)
        remote_monthly = remote_monthly.rename(columns={True: "remote_mentioned", False: "not_mentioned"})
        if "remote_mentioned" not in remote_monthly:
            remote_monthly["remote_mentioned"] = 0
        if "not_mentioned" not in remote_monthly:
            remote_monthly["not_mentioned"] = 0
        remote_monthly["remote_keyword_share_pct"] = (
            remote_monthly["remote_mentioned"] /
            (remote_monthly["remote_mentioned"] + remote_monthly["not_mentioned"]).replace(0, np.nan) * 100
        )
        display(remote_monthly.round(2))
        remote_monthly[["remote_mentioned", "not_mentioned"]].plot(
            figsize=(13, 6), marker="o")
        plt.title("Monthly Job Postings Mentioning Remote Work")
        plt.xlabel("Month")
        plt.ylabel("Number of postings")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
        remote_monthly[["remote_keyword_share_pct"]].plot(figsize=(13, 5), marker="o")
        plt.title("Share of Postings Mentioning Remote Work")
        plt.xlabel("Month")
        plt.ylabel("Remote-keyword share (%)")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
        remote_monthly.to_csv("task7_remote_work_trends.csv")
    else:
        print("Task 7 needs a title or description column.")
else:
    print("Task 7 skipped: valid publication dates are required.")
```

## Task 8: Predict future job-market trends and outline scenarios

This section combines the monthly category forecasts from Task 3 with simple scenario ranges. The scenarios are sensitivity assumptions (flat, moderate growth, moderate decline), not model outputs or confidence intervals. Use them to discuss possible outcomes transparently.

```python
if "monthly" in globals() and not monthly.empty:
    overall_monthly = monthly.groupby("month")["count"].sum().sort_index()
    if len(overall_monthly) >= 2:
        last_count = float(overall_monthly.iloc[-1])
        scenario_months = pd.date_range(
            pd.to_datetime(overall_monthly.index[-1]) + pd.offsets.MonthBegin(1),
            periods=6, freq="MS")
        scenario_rows = []
        for step, month in enumerate(scenario_months, start=1):
            scenario_rows.extend([
                {"month": month, "scenario": "Flat (0% monthly change)", "projected_postings": last_count},
                {"month": month, "scenario": "Growth (+5% monthly)", "projected_postings": last_count * (1.05 ** step)},
                {"month": month, "scenario": "Decline (-5% monthly)", "projected_postings": last_count * (0.95 ** step)}
            ])
        scenarios = pd.DataFrame(scenario_rows)
        display(scenarios.round(1))
        plt.figure(figsize=(12, 6))
        sns.lineplot(data=scenarios, x="month", y="projected_postings", hue="scenario", marker="o")
        plt.title("Six-Month Job Posting Scenarios (Illustrative Assumptions)")
        plt.xlabel("Month")
        plt.ylabel("Projected posting count")
        plt.tight_layout()
        plt.show()
        scenarios.to_csv("task8_future_market_scenarios.csv", index=False)
    else:
        print("Task 8 needs at least two observed months.")
else:
    print("Run Task 3 first; Task 8 uses the monthly category data created there.")
```
