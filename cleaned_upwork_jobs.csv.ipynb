# Step 1: Import the required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "all_upwork_jobs_2024-02-07-2024-03-24.csv"
)

print("Dataset loaded successfully!")
```

# Step 4: Check the number of rows and columns

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

# Step 5: Check column names

```python
print(df.columns)
```

```python
Clean the column names by removing extra spaces, converting them to lowercase, and replacing spaces with underscores.
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

print(df.columns)
```

# Step 6: Check data types
Each column should have the correct data type.

```python
print(df.info())
```

Check the data types individually:

```python
print(df.dtypes)
```

# Step 7: Check missing values

```python
print(df.isnull().sum())
```

```python
To calculate the percentage of missing values in each column:
```

```python
missing_percentage = (
    df.isnull().mean() * 100
).round(2)

print(missing_percentage)
```

# Step 8: Check duplicate records

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
You can also check whether job links are duplicated:
```

```python
print("Duplicate links:", df["link"].duplicated().sum())
```

If each link is intended to identify one job, you can retain only one record per link:

```python
df = df.drop_duplicates(
    subset=["link"],
    keep="first"
)
```

# Step 9: Handle missing values

```python
print(df[df["title"].isna()])
```

```python
Remove records without a job title or link, since they are important for identifying each job.
```

```python
df = df.dropna(
    subset=["title", "link"]
)
```

```python
df
```

```python
For the country column, you can label missing entries as Unknown:
```

```python
df["country"] = df["country"].fillna("Unknown")
```

```python
df
```

# Step 10: Clean text columns

```python
df["title"] = df["title"].str.strip()

df["country"] = df["country"].str.strip()
```

```python
df
```

```python
Replace empty text values with missing values:
```

```python
df["title"] = df["title"].replace("", np.nan)

df["country"] = df["country"].replace("", np.nan)
```

```python
df
```

```python
If you replace empty titles with missing values, remove those rows again:
```

```python
df = df.dropna(subset=["title"])
```

```python
df
```

# Step 11: Convert columns to the correct data types
Convert the date column into a proper datetime format.

```python
df["published_date"] = pd.to_datetime(
    df["published_date"],
    errors="coerce",
    utc=True
)
```

```python
df
```

```python
Convert the numerical columns:
```

```python
numeric_columns = [
    "hourly_low",
    "hourly_high",
    "budget"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )
```

```python
Convert is_hourly to a Boolean data type:
```

```python
df["is_hourly"] = df["is_hourly"].astype("boolean")
```

# Check the results:

```python
print(df.dtypes)
```

# Step 12: Check invalid values
A clean dataset should not contain negative budgets or negative hourly rates.

```python
print("Negative budgets:",
      (df["budget"] < 0).sum())

print("Negative hourly rates:",
      ((df["hourly_low"] < 0) |
       (df["hourly_high"] < 0)).sum())
```

```python
Check whether the minimum hourly rate is greater than the maximum hourly rate.
```

```python
invalid_rates = df[
    (df["hourly_low"] > df["hourly_high"]) &
    df["hourly_low"].notna() &
    df["hourly_high"].notna()
]

print(invalid_rates)
```

# Step 13: Check outliers
Outliers are unusually high or low values. They may be errors, but they may also represent genuine high-paying jobs.
Use a box plot to examine the budget column:

```python
df["budget"].dropna().plot(
    kind="box",
    title="Budget Outlier Detection"
)

plt.ylabel("Budget")
plt.show()
```

```python
You can use the Interquartile Range (IQR) method to identify unusually high budgets.
```

```python
budget = df["budget"].dropna()

Q1 = budget.quantile(0.25)
Q3 = budget.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["budget"] < lower_limit) |
    (df["budget"] > upper_limit)
]

print("Number of budget outliers:", len(outliers))
```

# Step 14: Create useful new columns

```python
df["average_hourly_rate"] = df[
    ["hourly_low", "hourly_high"]
].mean(axis=1)
```

```python
Create a month column:
```
